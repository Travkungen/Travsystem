```python
# ============================================================

# PHOENIX 15 — CELL 2

# READ ONLY: verifiera sparade Phoenix 15-filer

# ============================================================

import os

BASE = "/content/drive/MyDrive/PhoenixTrav"

paths = [

    f"{BASE}/phoenix_15_live",

    f"{BASE}/phoenix_15_core34_verified_baseline",

    f"{BASE}/PHOENIX15_MASTER_BACKUPS",

    f"{BASE}/bootstrap",

]

print("=" * 70)

print("PHOENIX 15 — SPARAD MILJÖ")

print("=" * 70)

for path in paths:

    print(f"\n📁 {path}")

    if os.path.exists(path):

        items = sorted(os.listdir(path))

        print(f"   Antal objekt: {len(items)}")

        for item in items[:40]:

            full = os.path.join(path, item)

            if os.path.isfile(full):

                size_mb = os.path.getsize(full) / (1024 * 1024)

                print(f"   📄 {item}  ({size_mb:.2f} MB)")

            else:

                print(f"   📂 {item}")

        if len(items) > 40:

            print(f"   ... {len(items)-40} objekt till")

    else:

        print("   ⚠️ Mappen finns inte")

print("\n" + "=" * 70)

print("DATABASE")

print("=" * 70)

db = f"{BASE}/phoenix_trav.db"

if os.path.exists(db):

    size_mb = os.path.getsize(db) / (1024 * 1024)

    print(f"✅ {db}")

    print(f"Storlek: {size_mb:.2f} MB")

else:

    print("❌ Databasen hittades inte")

print("\n⛔ READ ONLY — inget har ändrats.")

print("=" * 70)
```
